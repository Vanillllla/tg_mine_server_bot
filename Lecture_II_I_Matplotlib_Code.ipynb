```python
# Example 14
from random import randint
import matplotlib.pyplot as plt


X15 = range(100)
f = True
while True :
    Y15 = [x*randint(1, 10) for x in X15]
    Y16 = [x*randint(5, 10) for x in X15]
    s = []
    c = 0
    s = s + Y15
    s = s + Y16
    t = max(s)/100000
    for i in range(100):
        if Y15[i] - Y16[i] <= t:
            # print(Y15[i] - Y16[i], t, c)
            c += 1
    if c >= 90:
        break

fig, ax = plt.subplots()
ax.plot(X15, Y15, label='Max 10')
ax.plot(X15, Y16, label='Max 5')
fig.set_figheight(3)
fig.set_figwidth(6)
ax.legend()

plt.show()
```

```python
# Example 24
import matplotlib.pyplot as plt


X22 = range(1, 11)
Y22 = [x**2 for x in X22]
Y23 = [(x**2 + x) for x in X22]
Y24 = [(x**2 - x) for x in X22]
xerr22 = [0.05*x for x in X22]
yerr22 = [0.1*y for y in Y22]
yerr23 = [0.1*y for y in Y23]
xerr24 = [0.05*x for x in X22]

fig, ax = plt.subplots()
ax.errorbar(X22, Y22, xerr=xerr22, yerr=yerr22, marker='*', color='r', markersize=6, linestyle='none', linewidth=1, label='Both')
ax.errorbar(X22, Y23, yerr=yerr23, marker='o', color='b', markersize=3, linestyle='none', label='Y Error')
ax.errorbar(X22, Y24, xerr=xerr24, marker='x', color='g', markersize=5, linestyle='none', label='X Error')
ax.xaxis.set_tick_params(labelsize=14)
ax.yaxis.set_tick_params(labelsize=14)
ax.set_title('Errors Examples', fontsize=16)
ax.set_xlabel('Number', fontsize=16)
ax.set_ylabel('Value', fontsize=16)
ax.legend(loc='upper left')
fig.set_figheight(3)
fig.set_figwidth(5)
plt.show()
```

```python
# Example 25
import matplotlib.pyplot as plt


X22 = range(1, 101, 10)
Y22 = [x for x in X22]
xerr22 = [0.05*x for x in X22]
yerr22 = [0.2*y for y in Y22]
Y23 = [x**2 for x in X22]
xerr23 = [0.1*x for x in X22]
yerr23 = [0.3*y for y in Y23]
Y24 = [x**3 for x in X22]
xerr24 = [0.15*x for x in X22]
yerr24 = [0.4*y for y in Y24]


fig, ax = plt.subplots()
ax.errorbar(
    X22, Y22, xerr=xerr22, yerr=yerr22, marker='*', color=(0.1, 0.5, 0.9), markersize=8,
    linestyle='dotted', linewidth=1, elinewidth=1, capsize=3, label='X'
)
ax.errorbar(
    X22, Y23, xerr=xerr23, yerr=yerr23, marker='o', color=(0.9, 0.5, 0.1), markersize=4,
    linestyle='dashed', linewidth=1, elinewidth=1, capsize=3, label='X**2'
)
ax.errorbar(
    X22, Y24, xerr=xerr24, yerr=yerr24, marker='^', color=(0.5, 0.1, 0.9), markersize=6,
    linestyle='solid', linewidth=1, elinewidth=1, capsize=3, label='X**3'
)
ax.set_yscale('log')
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.set_title('Graph Title', fontname='monospace', fontsize=20)
ax.set_xlabel('Number', fontsize=18)
ax.set_ylabel('Value', fontsize=18)
ax.set_xticks(list(range(0, 100, 10)))
ax.ticklabel_format(axis='x', style='sci', scilimits=(0, 0))
ax.grid()
ax.grid(axis='y', which='minor', alpha=0.3)
ax.legend(loc='lower right')
fig.set_figheight(4)
fig.set_figwidth(6)
plt.show()
```
